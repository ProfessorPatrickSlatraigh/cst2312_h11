## Glossary

| Term | Meaning |
|:---|:---|
| Aliasing | Two or more names that refer to the same object. Changing a mutable object through one name is visible through the others. |
| Base case | The condition in a recursive function under which the function returns without calling itself. |
| Boolean mask | A Series of `True` and `False` values used to select rows from a DataFrame. |
| Class variable | A variable defined in the class body and shared by all instances of the class. |
| Greedy quantifier | A quantifier such as `+` or `*` that matches as much text as possible while still allowing the whole pattern to match. |
| Instance variable | A variable stored on one object, usually created in `__init__` through `self`. |
| Mutable | Describes an object that can be changed after creation, such as a list or a dictionary. |
| Overriding | Defining a method in a subclass with the same name as a method in the parent class, so that the subclass version is used. |
| Polymorphism | The ability of objects of different classes to respond to the same method call in class-specific ways. |
| Raw string | A string literal with the prefix `r`, in which backslashes are kept as literal characters. |
| Recursion | A technique in which a function solves a problem by calling itself on a smaller version of the problem. |
| Slice | A portion of a sequence selected with `[start:stop:step]`. The stop position is excluded. |

def is_palindrome(text):
    """Return True when text is a palindrome, ignoring case."""
